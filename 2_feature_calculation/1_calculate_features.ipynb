```python
import numpy as np
import pandas as pd
import gc
```

```python
#dataset = '../../../../0_apo_crbp2_2rcq/1_eq_prod/output_coords_res.csv' 
#dataset = '../../../../1_holo_crbp2_2rct_rtol/1_eq_prod/output_coords_res.csv' 
dataset = '../../../../8_holo_crbp2_7k3i_2lg/1_eq_prod/output_coords_res.csv' 
```

```python
### STEP 0. Load input data (coordinates)
gc.collect()
df = pd.read_csv(dataset)
df.dropna(how='all', axis=1, inplace=True)
varListCoord = df.columns.tolist()
nRes = int(len(varListCoord) / 3)
print("We have", nRes, "residues in total.")
```

```python
### STEP 1. Reshape coordinates into 3D NumPy arrays
coords = df.values.reshape(df.shape[0], nRes, 3)  # (n_samples, n_residues, 3)
```

```python
### STEP 2. Compute pairwise distances efficiently
pair_indices = [(x, y) for x in range(nRes) for y in range(x + 1, nRes)]
pair_names = [f"{varListCoord[3*x][:-2]}.{varListCoord[3*y][3:-2]}" for x, y in pair_indices]
```

```python
# Calculate distances using broadcasting
resid_differences = coords[:, np.newaxis, :, :] - coords[:, :, np.newaxis, :]
distances = np.sqrt(np.sum(resid_differences**2, axis=-1))
```

```python
# Extract upper triangle distances based on pair indices
dist_matrix = np.array([distances[:, x, y] for x, y in pair_indices]).T
```

```python
### STEP 3. Convert results to DataFrame and save
result_df = pd.DataFrame(dist_matrix, columns=pair_names)
#result_df.to_csv('output_dist_res_2rcq_apo.csv', index=False)
#result_df.to_csv('output_dist_res_2rct_rtol.csv', index=False)
result_df.to_csv('output_dist_res_7k3i_2lg.csv', index=False)
```
